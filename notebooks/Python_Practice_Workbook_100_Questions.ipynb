
### Question Q004 — Exponential Backoff Retry Schedule

**Difficulty:** Beginner &nbsp;|&nbsp; **Primary category:** Python fundamentals and control flow  
**Relevant roles:** AI Engineer, Data Engineer, Python Developer (data/AI systems)  
**Skills tested:** for loops, exponentiation, min(), round(), input validation  
**Estimated time:** 15 minutes

#### 1. Scenario

Calls to an external LLM API occasionally fail with rate-limit errors. The client retries using exponential backoff: each wait is longer than the last, but never longer than a cap. You need to compute the planned wait times up front so they can be logged and unit-tested.

#### 2. Problem statement

Write `backoff_schedule(base, factor, max_retries, cap)` that returns the list of delays (in seconds) before each retry. Retry *i* (starting at 0) waits `base * factor ** i` seconds, limited to at most `cap`. Round every delay to 3 decimals.

#### 3. Function signature

```python
def backoff_schedule(base: float, factor: float, max_retries: int, cap: float) -> list[float]:
```

#### 4. Input specification

- `base`: first delay in seconds, > 0.
- `factor`: growth multiplier, ≥ 1.
- `max_retries`: number of retries, an `int` ≥ 0.
- `cap`: maximum delay, ≥ `base`.

#### 5. Expected output

A `list[float]` with exactly `max_retries` delays, each rounded with `round(x, 3)`.

#### 6. Rules and constraints

- Delay *i* is `min(base * factor ** i, cap)`, rounded to 3 decimal places.
- Return `[]` when `max_retries` is 0.
- Raise `ValueError` if `base <= 0`, `factor < 1`, `max_retries < 0` or `cap < base`.

#### 7. Example input (capped growth)

```python
backoff_schedule(0.5, 2, 5, 4)
```

#### 8. Expected output

```python
[0.5, 1.0, 2.0, 4.0, 4.0]
```

#### 9. Additional examples

**rounding removes float noise**

```python
backoff_schedule(0.1, 3, 4, 10)
```

Expected output:

```python
[0.1, 0.3, 0.9, 2.7]
```

**boundary: no retries**

```python
backoff_schedule(1, 2, 0, 30)
```

Expected output:

```python
[]
```

#### 10. Acceptance criteria

- [ ] Produces the correct number of delays.
- [ ] Delays grow geometrically and are capped.
- [ ] Floating-point noise such as `0.30000000000000004` is removed by rounding.
- [ ] Invalid parameters raise `ValueError`.

### Question Q005 — Normalize Bounding-Box Coordinates

**Difficulty:** Beginner &nbsp;|&nbsp; **Primary category:** Python fundamentals and control flow

#### 1. Scenario

An object-detection labelling tool exports boxes as two opposite corners `(x1, y1, x2, y2)`, but annotators may drag from any corner, so `x1` can be larger than `x2`. The training pipeline expects `(x, y, width, height)` with `(x, y)` as the top-left corner.

#### 2. Problem statement

Write `normalize_bbox(box)` that converts two opposite corners in any order into `(x, y, width, height)`, where `x`/`y` are the smallest coordinates and width/height are non-negative.

- `x = min(x1, x2)`, `y = min(y1, y2)`, `width = abs(x2 - x1)`, `height = abs(y2 - y1)`.

**Difficulty:** Advanced &nbsp;|&nbsp; **Primary category:** Integrated data and AI engineering scenarios  
**Relevant roles:** AI Engineer, Applied AI Engineer, Python Developer (data/AI systems)  
**Skills tested:** recursive validation, path-tracking error messages, type dispatch, JSON-like data  
**Estimated time:** 45 minutes

#### 1. Scenario

An LLM extracts sentiment, confidence and entities from support tickets as JSON. Models occasionally invent fields, return confidence 1.2 or use unknown entity types. Before storing results you validate them against a small schema and log precise errors.

#### 2. Problem statement

Write `validate_llm_output(payload, schema)` that returns a sorted list of error messages describing every problem (empty list = valid).

#### 3. Function signature

```python
def validate_llm_output(payload, schema: dict) -> list[str]:
```

#### 4. Input specification

- `schema` nodes have a `type` of `object`, `array`, `string`, `number`, `integer` or `boolean`.
- object: `properties` (name → schema), optional `required` (list) and `additional` (bool, default `True`).
- array: `items` (schema), optional `min_items`. string: optional `enum`, `max_length`. number/integer: optional `minimum`, `maximum`.

#### 5. Expected output

`list[str]` sorted alphabetically. Each message is `'<path>: <problem>'`.

- Raise `ValueError` for an unknown schema type.

**Difficulty:** Expert &nbsp;|&nbsp; **Primary category:** Sorting, searching, and algorithmic problem-solving  
**Relevant roles:** AI Engineer, Data Scientist, Data Engineer  
**Skills tested:** set similarity (Jaccard), inverted index for candidate generation, union-find, deterministic output  
**Estimated time:** 60 minutes

#### 1. Scenario

Before building a RAG index you de-duplicate a scraped knowledge base: many pages are copies with tiny edits. Comparing every pair of documents is quadratic, so you only compare documents that share at least one word *shingle*, then group transitive matches.

#### 2. Problem statement

Write `cluster_near_duplicates(docs, threshold, shingle_size=3)` that returns groups of document ids whose shingle-set Jaccard similarity links them (directly or transitively).

#### 3. Function signature

```python
def cluster_near_duplicates(docs: dict[str, str], threshold: float, shingle_size: int = 3) -> list[list[str]]:
```

#### 4. Input specification

- `docs`: doc id → text.
- `threshold`: float in (0, 1].
- `shingle_size`: `int` ≥ 1.

#### 5. Expected output

A list of clusters (each a sorted list of ≥ 2 ids), sorted by their first id.

#### 6. Rules and constraints

- Tokens: `re.findall(r'[a-z0-9]+', text.lower())`. Shingles: the set of tuples of `shingle_size` consecutive tokens; a document with 1..shingle_size−1 tokens has the single shingle of all its tokens; no tokens → no shingles.

- Clusters are connected components of the link graph (transitive).

### Question Q091 — Point-in-Time Correct Feature Join (As-Of Join)

**Difficulty:** Expert &nbsp;|&nbsp; **Primary category:** Data processing and analytical transformations  
**Relevant roles:** Machine Learning Engineer, Data Scientist, Data Engineer  
**Skills tested:** avoiding label leakage, per-key sorting, bisect, staleness limits, tie semantics  
**Estimated time:** 50 minutes

#### 1. Scenario

Your churn model looked great offline and failed in production. The cause: training rows were joined with features computed *after* the label event (data leakage). The fix is an as-of join: each label gets the latest feature values known at the label's timestamp.

#### 2. Problem statement

Write `asof_join(labels, features, max_staleness=None)` that attaches to each label the most recent feature row for the same entity with `feature_ts <= label_ts`.

#### 3. Function signature

```python
def asof_join(labels: list[dict], features: list[dict], max_staleness: int | None = None) -> list[dict]:
```

#### 4. Input specification

- `labels`: dicts with `entity_id`, `ts` (int seconds) and other fields.
- `features`: dicts with `entity_id`, `ts` and `features` (a dict). Unsorted.
- `max_staleness`: optional int ≥ 0.

#### 5. Expected output

For each label (input order): a copy of the label plus `features` (a copy of the matched dict, or `None`) and `feature_ts` (int or `None`).

#### 6. Rules and constraints

- Only feature rows with `ts <= label ts` may be used (equal timestamps allowed).
- If several feature rows share the winning timestamp, the one appearing later in `features` wins.

- Do not modify the inputs. Raise `ValueError` for negative `max_staleness`.

#### 2. Problem statement

Write `run_batch_inference(items, predict_batch, batch_size, max_retries=2, retryable=(TimeoutError, ConnectionError))` that returns predictions for every item that can be scored and a failure reason for every item that cannot.

#### 3. Function signature

```python
def run_batch_inference(items: list[dict], predict_batch, batch_size: int, max_retries: int = 2,

```

#### 4. Input specification

#### 2. Problem statement

Write `assemble_stream(events)` that validates the event sequence and returns the assembled message.

#### 3. Function signature

```python
def assemble_stream(events: list[dict]) -> dict:
```

#### 4. Input specification

- `text_delta` with `text`; `tool_call_delta` with `call_id`, `arguments` (a JSON fragment) and – on any one delta of that call – `name`;

### Question Q099 — Pack Retrieved Context into a Token Budget

**Difficulty:** Expert &nbsp;|&nbsp; **Primary category:** Integrated data and AI engineering scenarios  
**Relevant roles:** AI Engineer, Applied AI Engineer, Machine Learning Engineer  
**Skills tested:** greedy selection with tie-breaking, normalised de-duplication, budget accounting, order restoration  
**Estimated time:** 45 minutes

#### 1. Scenario

Your RAG service retrieves ~50 passages per question, but the model's context window leaves room for only a fraction of them after reserving space for the answer. Retrieval also returns near-identical copies of the same passage. You need a deterministic packing step that keeps the most relevant unique passages and preserves their original order.

#### 2. Problem statement

Write `pack_context(docs, budget, reserve=0)` that selects passages to fit the token budget and builds the prompt context.

#### 3. Function signature

```python
def pack_context(docs: list[dict], budget: int, reserve: int = 0) -> dict:
```

#### 4. Input specification

- `docs`: dicts with unique `id`, `text` and numeric `score` (higher is more relevant), in retrieval order.
- `budget`, `reserve`: ints.

#### 5. Expected output

`{'selected': [ids in original order], 'used_tokens': int, 'dropped': [sorted ids], 'prompt': str}` where `prompt` joins `'[<id>]\n<text.strip()>'` for the selected docs with blank lines (`'\n\n'`).

#### 6. Rules and constraints

- A doc costs `len(text.split()) + 4` tokens (4 for the header and separator). Available tokens = `budget - reserve`.
- De-duplicate first: texts equal after lower-casing and collapsing whitespace are duplicates; keep the highest score (earlier doc on equal scores).
- Greedy: consider docs by score descending, then cost ascending, then id ascending; add a doc if it still fits, otherwise skip it and continue.
- Raise `ValueError` if `budget - reserve <= 0`, a score is not a number, or ids are not unique.

```python
     {'id': 'd3', 'text': 'Alpha  beta gamma delta', 'score': 0.7},
     {'id': 'd4', 'text': 'short text', 'score': 0.7},
     {'id': 'd5',
      'text': 'x x x x x x x x x x x x x x x x x x x x x x x x x x x x x x ',
      'score': 0.95}],

)
```

#### 8. Expected output

### Question Q100 — Sentence-Aware Text Chunking with Overlap for RAG

**Difficulty:** Expert &nbsp;|&nbsp; **Primary category:** Integrated data and AI engineering scenarios  
**Relevant roles:** AI Engineer, Applied AI Engineer, Data Engineer  
**Skills tested:** regex sentence splitting, greedy packing, overlap with progress guarantees, careful specification  
**Estimated time:** 60 minutes

#### 1. Scenario

Your document ingestion pipeline splits knowledge-base articles into chunks before embedding them. Chunks that cut sentences in half retrieve poorly, and without overlap, facts that span a chunk boundary are lost. The chunker must also never loop forever on awkward inputs such as very long sentences.

#### 2. Problem statement

Write `chunk_text(text, max_words, overlap_words)` that packs whole sentences into chunks of at most `max_words` words, repeating trailing sentences of the previous chunk as overlap.

#### 3. Function signature

```python
def chunk_text(text: str, max_words: int, overlap_words: int) -> list[dict]:
```

#### 4. Input specification

- `text`: any string.
- `max_words`: `int` ≥ 1.
- `overlap_words`: `int` with `0 <= overlap_words < max_words`.

#### 5. Expected output

A list of `{'index': i, 'text': str, 'word_count': int, 'unit_range': (start, end)}` – units `start..end-1` joined by single spaces.

#### 6. Rules and constraints

- Units: split the whitespace-normalised text with `re.split(r'(?<=[.!?])\s+', ...)`; a sentence with more than `max_words` words becomes several units of `max_words` words (the last may be shorter). Empty text → `[]`.
- Chunk 0 starts at unit 0. Every chunk greedily adds the following units while the total stays ≤ `max_words`.
- Each later chunk starts with the longest run of trailing units of the previous chunk whose total is ≤ `overlap_words`; if that overlap plus the next new unit exceeds `max_words`, drop overlap units from the front until it fits (possibly all of them).
- Stop once the last unit has been included. Raise `ValueError` for invalid `max_words`/`overlap_words`.

#### 7. Example input (sentence packing with overlap)

```python

    ('RAG systems retrieve context. They split documents into chunks. Each chunk '
     'should be self-contained! Overlap preserves continuity between chunks? Yes.'),
    10,
    5,
)
```
